##### Calculate total orders and total sales revenue.


```python
total_orders_count = "SELECT COUNT(*) AS total FROM orders;"
df = pd.read_sql(total_orders_count, con=engine)

# Fetch total revenue by joining order_items and orders (Fixed column name issue)
total_sales_revenue = """
SELECT SUM(oi.selling_price * oi.quantity) AS total_revenue 
FROM order_items oi
INNER JOIN orders o ON oi.order_id = o.order_id
WHERE o.order_status = 'Completed';
"""
df2 = pd.read_sql(total_sales_revenue, con=engine)

# Extract single numeric values safely
count_value = int(df.iloc[0, 0])
revenue_value = float(df2.iloc[0, 0]) if pd.notnull(df2.iloc[0, 0]) else 0.0

print(f"Total Orders:        {count_value:,}")
print(f"Total Sales Revenue: ₹{revenue_value:,.2f}")
```

##### Find the top five products by quantity sold.


```python
top_products = """
SELECT p.product_name, CAST(SUM(oi.quantity) AS SIGNED) AS total_quantity_sold FROM
order_items oi 
INNER JOIN products p ON p.product_id = oi.product_id
GROUP BY p.product_name
ORDER BY total_quantity_sold DESC LIMIT 5;"""

df = pd.read_sql(top_products, con=engine)
format_table(df, "Top Products")
```

Add a new column : Revenue on orders_items table


```python
# SQL to add the column and populate it with data
alter_queries = [
    "ALTER TABLE order_items ADD COLUMN revenue DECIMAL(10, 2);",
    "UPDATE order_items SET revenue = quantity * selling_price;",
]

with engine.begin() as conn:
    for query in alter_queries:
        conn.execute(text(query))

print("New column 'revenue' created and populated successfully!")

query = "SELECT * FROM order_items LIMIT 5;"
df = pd.read_sql(query, con=engine)

cols_to_convert = ["order_item_id", "order_id", "product_id", "quantity"]
df[cols_to_convert] = df[cols_to_convert].astype("Int64")

format_table(df, "Order Items")
```

##### Calculate revenue by product category.


```python
revenue_category = """
SELECT p.category, CAST(SUM(oi.revenue) AS SIGNED) AS 'Total Revenue' FROM order_items oi
INNER JOIN products p ON p.product_id = oi.product_id
GROUP BY p.category;
"""

df = pd.read_sql(revenue_category, con=engine)
format_table(df, "Revenue by Product Category")
```

##### Calculate total revenue and estimated profit by product.


```python
profit_query = """
SELECT p.product_name,
SUM(oi.revenue) AS total_revenue,
SUM(oi.revenue - (oi.quantity * p.unit_cost)) AS estimated_profit
FROM order_items oi
INNER JOIN products p ON p.product_id = oi.product_id
GROUP BY p.product_id, p.product_name
ORDER BY estimated_profit DESC;
"""
df_profit = pd.read_sql(profit_query, con=engine)

format_table(df_profit, "Product Performance & Profitability")
```

##### Find monthly revenue and compare it across months.


```python
monthly_revenue_query = """
SELECT 
    DATE_FORMAT(o.order_date, '%Y-%m') AS order_month,
    SUM(oi.revenue) AS monthly_revenue
FROM order_items oi
INNER JOIN orders o ON oi.order_id = o.order_id
WHERE o.order_status = 'Completed'
GROUP BY DATE_FORMAT(o.order_date, '%Y-%m')
ORDER BY order_month ASC;
"""
df_monthly = pd.read_sql(monthly_revenue_query, con=engine)
format_table(df_monthly, "Monthly Revenue Summary")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5), dpi=100)
plt.style.use("seaborn-v0_8-whitegrid")  # Standard clean grid style

plt.plot(
    df_monthly["order_month"],
    df_monthly["monthly_revenue"],
    marker="o",
    linewidth=2.5,
    color="#1f77b4",
    label="Revenue",
)

plt.title("Monthly Revenue Progression Trend", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Billing Month (YYYY-MM)", fontsize=11, labelpad=10)
plt.ylabel("Revenue (₹)", fontsize=11, labelpad=10)

current_values = plt.gca().get_yticks()
plt.gca().set_yticklabels([f"₹{x:,.0f}" for x in current_values])  # Formats as ₹10,000

plt.tight_layout()
plt.show()
```

##### Find the top five customers by total spending.


```python
top_customer = """
SELECT c.customer_name, SUM(oi.revenue) AS total_spending
FROM order_items oi
INNER JOIN orders o ON o.order_id = oi.order_id
INNER JOIN customers c ON c.customer_id = o.customer_id
WHERE o.order_status = 'Completed'
GROUP BY c.customer_id, c.customer_name
ORDER BY total_spending DESC
LIMIT 5;
"""

df = pd.read_sql(top_customer, con=engine)
format_table(df, "Top 5 Customers by Spending")
```

##### Identify customers who have never placed an order.


```python
noValue_customer = """
SELECT c.customer_name FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
"""

df = pd.read_sql(noValue_customer, con=engine)
format_table(df, "Customers with No Orders")
```

##### Calculate the number of orders per customer.


```python
orders_per_customer = """
SELECT c.customer_name, COUNT(o.order_id) AS total_orders
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
ORDER BY total_orders DESC;
"""

df = pd.read_sql(orders_per_customer, con=engine)
format_table(df, "Orders Per Customer")
```

##### Find repeat customers who placed more than one completed order.


```python
repeat_customer = """
SELECT c.customer_name, COUNT(o.order_id) AS total_orders
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id AND o.order_status = 'Completed'
GROUP BY c.customer_id, c.customer_name
HAVING total_orders > 1
ORDER BY total_orders DESC;
"""

df = pd.read_sql(repeat_customer, con=engine)
format_table(df, "Repeat Customer List")
```

##### Calculate average order value (AOV).


```python
aov_query = """
SELECT SUM(oi.revenue) AS total_revenue,
COUNT(DISTINCT o.order_id) AS total_orders,
SUM(oi.revenue) / COUNT(DISTINCT o.order_id) AS average_order_value
FROM order_items oi
INNER JOIN orders o ON oi.order_id = o.order_id
WHERE o.order_status = 'Completed';
"""

df_aov = pd.read_sql(aov_query, con=engine)

revenue = float(df_aov.iloc[0, 0])
orders = int(df_aov.iloc[0, 1])
aov = float(df_aov.iloc[0, 2])

print(f"Total Revenue:  ₹{revenue:,.2f}")
print(f"Total Orders:   {orders:,}")
print(f"Grand AOV:      ₹{aov:,.2f}")
```

### Phase 4 — Business insights


##### Find product categories with the highest estimated profit.


```python
profit_category = """
SELECT p.category,
SUM(oi.revenue) AS total_revenue,
SUM(oi.revenue - (oi.quantity * p.unit_cost)) AS estimated_profit
FROM order_items oi
INNER JOIN products p ON p.product_id = oi.product_id
GROUP BY p.category
ORDER BY estimated_profit DESC;
"""

df = pd.read_sql(profit_category, con=engine)
format_table(df, "Profit by Product category")
```

##### Calculate each category's percentage contribution to revenue.


```python
category_contribution_query = """
SELECT p.category,
SUM(oi.revenue) AS category_revenue,
(SUM(oi.revenue) / (SELECT SUM(revenue) FROM order_items)) * 100 AS percentage_contribution
FROM order_items oi
INNER JOIN products p ON oi.product_id = p.product_id
GROUP BY p.category
ORDER BY percentage_contribution DESC;
"""

df_category = pd.read_sql(category_contribution_query, con=engine)
format_table(df_category, "Category Revenue Contribution")
```

##### Identify products that have never been ordered.


```python
faltu_product = """
SELECT 
    p.product_id,
    p.product_name,
    p.category,
    p.unit_price
FROM products p
LEFT JOIN order_items oi ON p.product_id = oi.product_id
WHERE oi.order_item_id IS NULL;
"""

df_unordered = pd.read_sql(faltu_product, con=engine)

if df_unordered.empty:
    print(
        "🎉 Great news! There is no dead stock. Every product in the catalog has been ordered at least once!"
    )
else:
    format_table(df_unordered, "Products with No Order History")
```

##### Compare completed, pending, and cancelled order counts.


```python
status_comparison_query = """
SELECT 
    order_status,
    COUNT(order_id) AS order_count
FROM orders
GROUP BY order_status
ORDER BY order_count DESC;
"""

df_status = pd.read_sql(status_comparison_query, con=engine)

format_table(df_status, "Order Status Comparison")
```

##### Find customers whose total spending is above the average customer spending.


```python
vip_customers_query = """
SELECT 
    c.customer_name,
    SUM(oi.revenue) AS total_spending
FROM order_items oi
INNER JOIN orders o ON o.order_id = oi.order_id
INNER JOIN customers c ON c.customer_id = o.customer_id
WHERE o.order_status = 'Completed'
GROUP BY c.customer_id, c.customer_name
HAVING total_spending > (
    -- Subquery: Calculates global average spending per customer
    SELECT AVG(customer_total)
    FROM (
        SELECT SUM(sub_oi.revenue) AS customer_total
        FROM order_items sub_oi
        INNER JOIN orders sub_o ON sub_o.order_id = sub_oi.order_id
        WHERE sub_o.order_status = 'Completed'
        GROUP BY sub_o.customer_id
    ) AS customer_spending_baselines
)
ORDER BY total_spending DESC;
"""

df_vip = pd.read_sql(vip_customers_query, con=engine)

if df_vip.empty:
    print("No customers are currently spending above the average baseline.")
else:
    format_table(df_vip, "Above-Average Spending Customers")
```
